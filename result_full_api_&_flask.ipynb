1 What is a RESTful API?


RESTful API is a way for systems to communicate over the internet. It follows REST (Representational State Transfer) principles using HTTP methods (like GET, POST) to perform operations like retrieving or sending data.


2 What is the concept of API specification?


An API specification is a document that describes how an API works, including available endpoints, required parameters, and responses. It’s like a user manual for developers.




 3 What is Flask, and why is it popular for building APIs?

Flask is a lightweight Python framework for building web applications and APIs. It’s popular because it’s simple, flexible, and has many extensions for added functionality.




4 What is routing in Flask?


Routing in Flask maps URLs (e.g., /home) to specific functions in your code. It decides what happens when a user visits a particular URL.


5 How do you create a simple Flask application?

```python
from flask import Flask
app = Flask(__name__)

@app.route('/')
def home():
    return 'Hello, World!'

if __name__ == '__main__':
    app.run()
```

6 What are HTTP methods used in RESTful APIs?


HTTP methods tell the API what action to perform:


GET: Retrieve data.


POST: Send new data.


PUT: Update existing data.


DELETE: Remove data.

8 What is the difference between GET and POST HTTP methods?


GET: Used to request data (e.g., view a webpage). Data is sent in the URL.


POST: Used to send data (e.g., submit a form). Data is sent in the request body.

1 How do you create a basic Flask application?


```python
from flask import Flask
app = Flask(__name__)


@app.route('/')
def home():
    return 'Welcome to Flask!'


if __name__ == '__main__':
    app.run()
```

2 How do you serve static files like images or CSS in Flask?


```python
<link rel="stylesheet" href="{{ url_for('static', filename='styles.css') }}">
```

3 How do you define different routes with different HTTP methods in Flask?


```python
@app.route('/submit', methods=['GET', 'POST'])
def handle_request():
    if request.method == 'POST':
        return 'You sent a POST request!'
    return 'You sent a GET request!'
```

4 How do you render HTML templates in Flask?




```python
from flask import render_template
@app.route('/')
def home():
    return render_template('index.html')


```

5 How can you generate URLs for routes in Flask using url_for?


```python
from flask import url_for
@app.route('/about')
def about():
    return 'About page'


@app.route('/')
def home():
    return f'<a href="{url_for("about")}">Go to About</a>'
```

6 How do you handle forms in Flask?

```python
<!-- HTML form -->
<form action="/submit" method="POST">
    <input type="text" name="username">
    <button type="submit">Submit</button>
</form>

@app.route('/submit', methods=['POST'])
def submit_form():
    username = request.form['username']
    return f'Hello, {username}!'


```

7 How can you validate form data in Flask?


```python
@app.route('/submit', methods=['POST'])
def submit_form():
    username = request.form['username']
    if not username:
        return 'Username is required!', 400
    return f'Hello, {username}!'


```

8 How do you manage sessions in Flask?




```python
from flask import session
app.secret_key = 'your_secret_key'


@app.route('/set')
def set_session():
    session['username'] = 'John'
    return 'Session set!'


@app.route('/get')
def get_session():
    return f'Hello, {session.get("username", "Guest")}'
```

9 How do you redirect to a different route in Flask?




```python
from flask import redirect
@app.route('/old')
def old_route():
    return redirect('/new')


@app.route('/new')
def new_route():
    return 'This is the new route!'


```

10 How do you handle errors in Flask (e.g., 404)?


```python
@app.errorhandler(404)
def page_not_found(error):
    return 'Page not found!', 404


```

11 How do you structure a Flask app using Blueprints?


```python
from flask import Blueprint


user_bp = Blueprint('user', __name__)


@user_bp.route('/profile')
def profile():
    return 'User Profile'


app.register_blueprint(user_bp, url_prefix='/user')


```

12 How do you define a custom Jinja filter in Flask?


```python
@app.template_filter('reverse')
def reverse_string(s):
    return s[::-1]


```

13 How can you redirect with query parameters in Flask?


```python
@app.route('/old')
def old_route():
    return redirect(url_for('new_route', param='value'))


@app.route('/new')
def new_route():
    param = request.args.get('param')
    return f'Received param: {param}'
```
